always pick median over mean when trying tro replace missing values because median is robust to outliers preventing skewing of results.

fill the missing with the median review rating within each category. that is it'll be replaced the median value of something within its category

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median())) ## fill the missing values with the median review rating within each category
```

create a column age_group

```python
#create a column age_group
labels = ['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels) ## create a column age_group using quantile cut
```

create purchase frequency days column

the current sheet shows how often customers shop but its in text format. it needs to be converted into a numerical format so that i can perform calculations on it. i.e number of days btw purchases


```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Biweekly': 14,
    'Annually': 365,
    'Every 3 Months': 90,
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping) ## create a new column purchase_frequency_days by mapping
#the frequency_of_purchases column to the frequency_mapping dictionary)
```
